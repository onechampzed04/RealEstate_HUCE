```python
import numpy as np 
import pandas as pd
```

```python
import matplotlib as plt
import seaborn as sns
import matplotlib.pyplot as plt2
```

```python
data_df = pd.read_csv("vietnam_housing_dataset.csv")
```

```python
data_df.info()
```

```python
for col in data_df.columns:
    missing_data = data_df[col].isna().sum()
    missing_percent = missing_data/len(data_df)*100
    print(f"column {col} have {missing_percent}% missing data")
```

```python
data_df.head()
```

```python
fig, ax = plt2.subplots(figsize=(8,5))
sns.heatmap(data_df.isna(), cmap="Blues", cbar=False, yticklabels=False)
```

```python
pd.set_option('display.max_colwidth', None)
data_address = data_df["Address"]
print(data_address) 
```

```python
result = data_address.str.split(r',', expand= True)
# Tắt tính năng tự động ngắt dòng khi hiển thị
pd.set_option('display.expand_frame_repr', False)

# Nếu cột vẫn bị khuất do quá dài, bạn có thể tăng chiều rộng hiển thị tối đa
pd.set_option('display.max_colwidth', None)

print(result)
```

```python
for col in result.columns:
    missing_data = result[col].isna().sum()
    missing_percent = missing_data/len(result)*100
    print(f"col {col} has {missing_percent}% missing value is nan")
```

```python
# Kiểm tra nội dung cột 4 (những dòng hiếm hoi có dữ liệu)
print("Dữ liệu ở cột 4:")
print(result[result[4].notna()][4].unique())

# Kiểm tra nội dung cột 5
print("\nDữ liệu ở cột 5:")
print(result[result[5].notna()][5].unique())
```
